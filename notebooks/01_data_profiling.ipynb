## Initial Data Profiling Findings

The Online Retail dataset contains **541,909 transaction-line records across 8 columns**. The dataset covers transaction, product, customer, and geographic information.

### Key Findings

1. **Transaction-level structure**
   The dataset contains transaction-line records rather than one row per order. There are 25,900 unique invoice numbers, indicating that a single invoice can contain multiple product lines.

2. **Customer identification**
   `CustomerID` contains 406,829 non-null values out of 541,909 records. Approximately **24.93% of records do not have a CustomerID**. This is an important data quality issue because customer identification is required for customer-level analysis and RFM segmentation.

3. **Missing product descriptions**
   `Description` contains 1,454 missing values, representing approximately **0.27% of all records**.

4. **Transaction dates**
   `InvoiceDate` is already stored as a datetime data type. The observed transaction period ranges from December 2010 to December 2011.

5. **Quantity anomalies**
   `Quantity` contains negative values and extreme observations. The minimum value is -80,995 and the maximum is 80,995. These records require further investigation because negative quantities may represent cancellations or returns rather than data-entry errors.

6. **Unit price anomalies**
   `UnitPrice` contains negative and highly extreme values. The median unit price is 2.08, while the maximum reaches 38,970. These observations require further investigation before determining an appropriate treatment.

7. **Geographic concentration**
   The dataset contains 38 countries, but the majority of transaction records are associated with the United Kingdom. This indicates a strong concentration of business activity in the UK market.

### Initial Business Implications

The dataset is suitable for customer-level analysis and RFM segmentation, but several data quality considerations must be addressed before analysis. In particular, missing `CustomerID` values, negative quantities, unusual unit prices, and transaction cancellations need to be investigated before creating the final analytical dataset.

## 4. Missing Value Analysis

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": df.isna().mean() * 100
})

missing_summary = missing_summary.sort_values(
    "missing_count",
    ascending=False
)

missing_summary["missing_percentage"] = (
    missing_summary["missing_percentage"].round(2)
)

missing_summary
```

### Missing Value Findings

The dataset contains missing values in two columns: `CustomerID` and `Description`.

* `CustomerID` has **135,080 missing values (24.93%)**, making it the most significant missing-data issue in the dataset. Since customer identification is required for customer-level analysis and RFM segmentation, records without a `CustomerID` cannot be reliably assigned to a customer.
* `Description` has **1,454 missing values (0.27%)**. The missing proportion is relatively small, and `StockCode` is complete across all records, so these records require further investigation before deciding how they should be treated.
* The remaining six columns contain no missing values.

### Business Implication

Missing `CustomerID` values may reduce the number of transactions available for customer-level analysis. However, these records should not be removed during the profiling stage. Their relationship with transaction cancellations, product information, and other transaction characteristics should first be investigated during the data cleaning phase.

```python
customer_missing = df["CustomerID"].isna()

cancelled = df["InvoiceNo"].astype(str).str.startswith("C")

comparison = pd.DataFrame({
    "CustomerID_Status": [
        "Missing",
        "Available"
    ],
    "Rows": [
        customer_missing.sum(),
        (~customer_missing).sum()
    ],
    "Cancelled_Rows": [
        (customer_missing & cancelled).sum(),
        ((~customer_missing) & cancelled).sum()
    ]
})

comparison
```

```python
comparison["Cancellation_Rate"] = (
    comparison["Cancelled_Rows"] /
    comparison["Rows"] * 100
).round(2)

comparison
```

```python
df.groupby(
    df["CustomerID"].isna()
)["Quantity"].agg(
    ["count", "mean", "median", "min", "max"]
)
```

### CustomerID Missingness Investigation

Further investigation shows that transactions with and without `CustomerID` have different characteristics.

Transactions with an available `CustomerID` have a cancellation rate of **2.19%**, compared with **0.28%** for transactions without a `CustomerID`. This indicates that missing customer identification is not evenly distributed across transaction types.

The quantity distribution also differs substantially. Transactions with an available `CustomerID` have a median quantity of **5 units**, while transactions without a `CustomerID` have a median quantity of **1 unit**. Extreme quantity values are also more pronounced among records with an identified customer.

Geographically, transactions with an available `CustomerID` span **37 countries**, while transactions without a `CustomerID` span only **9 countries**.

These findings suggest that records with missing `CustomerID` may have different transaction characteristics. Therefore, missing `CustomerID` records should be investigated further rather than treated as random missing data.

For customer-level analysis and RFM segmentation, only transactions that can be reliably associated with a customer will be eligible for customer-level aggregation. However, the raw records will remain unchanged until the data cleaning phase.

```python
missing_customer_country = (
    df.loc[df["CustomerID"].isna(), "Country"]
    .value_counts()
)

missing_customer_country
```

#### Missing CustomerID by Country

Missing `CustomerID` values are highly concentrated geographically.

The United Kingdom contributes **133,600 missing CustomerID records**, representing the vast majority of all missing customer identifiers. However, the missing rate within the UK is **26.96%**, indicating that more than one-quarter of UK transaction records cannot be directly linked to a customer.

Hong Kong shows a different pattern: all **288 transaction records** from Hong Kong have a missing `CustomerID`, resulting in a **100% missing rate**.

The `Unspecified` country category also has a relatively high missing rate of **45.29%**.

These patterns indicate that missing `CustomerID` values are not randomly distributed across countries. This is important when interpreting customer-level analysis because the resulting customer population may not fully represent all transaction activity.

For RFM and customer segmentation, transactions without a valid `CustomerID` will not be eligible for customer-level aggregation. However, these records will remain available for transaction-level analysis where appropriate.

### Duplicate Row Findings

The dataset contains **5,268 exact duplicate rows**, representing **0.97% of all records**. These duplicate rows are distributed across **1,933 invoices**.

Duplicate detection was performed across all eight columns, meaning that a record is considered an exact duplicate only when all transaction attributes are identical.

## 6. Outlier Analysis

```python
Q1_quantity = df["Quantity"].quantile(0.25)
Q3_quantity = df["Quantity"].quantile(0.75)

IQR_quantity = Q3_quantity - Q1_quantity

lower_quantity = Q1_quantity - 1.5 * IQR_quantity
upper_quantity = Q3_quantity + 1.5 * IQR_quantity

print(f"Q1       : {Q1_quantity}")
print(f"Q3       : {Q3_quantity}")
print(f"IQR      : {IQR_quantity}")
print(f"Lower    : {lower_quantity}")
print(f"Upper    : {upper_quantity}")
```

```python
quantity_outliers = df[
    (df["Quantity"] < lower_quantity) |
    (df["Quantity"] > upper_quantity)
]

print(f"Outlier rows: {len(quantity_outliers):,}")
print(
    f"Outlier percentage: "
    f"{len(quantity_outliers) / len(df) * 100:.2f}%"
)
```

```python
quantity_outliers["Quantity"].describe()
```

```python
quantity_outliers = df[
    (df["Quantity"] < lower_quantity) |
    (df["Quantity"] > upper_quantity)
].copy()

quantity_outliers["is_cancelled"] = (
    quantity_outliers["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
)

quantity_outliers["outlier_type"] = np.where(
    quantity_outliers["Quantity"] < lower_quantity,
    "Negative",
    "Positive"
)
```

```python
quantity_outliers.groupby(
    ["outlier_type", "is_cancelled"]
)["Quantity"].agg(
    ["count", "sum", "mean", "median", "min", "max"]
)
```

```python
quantity_outliers[
    quantity_outliers["Quantity"] > 100
]["Quantity"].describe()
```

```python
quantity_outliers[
    quantity_outliers["Quantity"] > 1000
][[
    "InvoiceNo",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "UnitPrice",
    "CustomerID",
    "Country"
]].sort_values(
    "Quantity",
    ascending=False
).head(20)
```

```python
negative_non_cancelled = df[
    (df["Quantity"] < lower_quantity) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
]

print(
    f"Negative outliers without cancellation prefix: "
    f"{len(negative_non_cancelled):,}"
)

negative_non_cancelled[
    [
        "InvoiceNo",
        "StockCode",
        "Description",
        "Quantity",
        "InvoiceDate",
        "UnitPrice",
        "CustomerID",
        "Country"
    ]
].sort_values(
    "Quantity"
).head(20)
```

### Quantity Outlier Findings

Using the IQR method, the lower and upper bounds for `Quantity` were calculated as **-12.5** and **23.5**, respectively. This identified **58,619 rows (10.82%)** as statistical outliers.

However, the statistical definition alone is not sufficient to determine whether these records are invalid. Further investigation revealed several distinct business patterns.

First, 56,635 positive outlier records were identified. Many of these represent high-volume purchases and may be legitimate bulk transactions. Therefore, high positive quantities will not automatically be removed.

Second, 1,984 negative quantity outliers were identified. Of these, 1,081 records are associated with invoices beginning with `C`, indicating cancellation or return-related transactions. Another 903 negative outliers do not have the `C` prefix. Their descriptions include terms such as "thrown away", "unsaleable", "destroyed", "mouldy", and "incorrect stock entry", indicating that these records are more likely to represent inventory adjustments, damaged goods, disposal, or stock corrections rather than customer cancellations.

Third, nine transactions were identified with both `Quantity > 1,000` and `UnitPrice = 0`. Several of these records have missing CustomerID or descriptions containing operational notes such as "incorrectly credited", "came coded as", or "?". These records are strong candidates for non-sales or data-quality records.

Therefore, IQR-based filtering will **not** be used to automatically remove Quantity outliers. Instead, transaction records will be classified based on their business context during the data cleaning stage. Legitimate high-volume purchases will be preserved, while cancellation and non-sales inventory adjustment records will be handled separately.

```python
Q1_price = df["UnitPrice"].quantile(0.25)
Q3_price = df["UnitPrice"].quantile(0.75)

IQR_price = Q3_price - Q1_price

lower_price = Q1_price - 1.5 * IQR_price
upper_price = Q3_price + 1.5 * IQR_price

print(f"Q1       : {Q1_price}")
print(f"Q3       : {Q3_price}")
print(f"IQR      : {IQR_price}")
print(f"Lower    : {lower_price}")
print(f"Upper    : {upper_price}")
```

```python
price_outliers = df[
    (df["UnitPrice"] < lower_price) |
    (df["UnitPrice"] > upper_price)
].copy()

print(f"Outlier rows: {len(price_outliers):,}")
print(
    f"Outlier percentage: "
    f"{len(price_outliers) / len(df) * 100:.2f}%"
)
```

### UnitPrice Outlier Findings

The IQR method identified an upper bound of **8.45** and classified **39,627 rows (7.31%)** as statistical outliers.

Further investigation showed that these outliers cannot be automatically considered invalid transactions.

Two records have negative `UnitPrice` values of **-11,062.06**. Both records have the description `Adjust bad debt`, missing CustomerID, and are not cancellations. These records represent accounting adjustments rather than normal product sales.

There are also **1,036 records with UnitPrice greater than 100**. Several of the most extreme records correspond to `AMAZONFEE`, `POSTAGE`, `Manual`, and other operational transactions. Therefore, extreme prices may represent fees or accounting-related transactions rather than product prices.

A total of **2,515 records have UnitPrice = 0**. Of these, 2,475 records (98.4%) have missing CustomerID. Many also have missing descriptions, suggesting that a substantial portion may represent non-sales or incomplete transaction records. However, zero-priced transactions should not automatically be removed because they may also represent legitimate free items, promotional items, or other business adjustments.

Therefore, IQR-based filtering will not be used to remove UnitPrice outliers. Instead, transaction-type indicators and business rules will be used to distinguish normal product sales from cancellations, fees, accounting adjustments, and other non-sales records.

## Business Interpretation

The profiling process shows that the dataset contains not only normal product sales, but also cancellations, returns, operational adjustments, fees, and other non-standard transactions.

### 1. Customer Analysis Limitation

Approximately **24.93% of transactions do not have a CustomerID**.

This means these transactions cannot be reliably attributed to a specific customer. Therefore, they should not be included in customer-level analysis such as:

- RFM analysis
- Customer segmentation
- Customer purchase frequency
- Customer monetary value
- Customer retention analysis

However, transactions without CustomerID may still contain valid information for overall sales and product analysis.

---

### 2. Transaction Quality

The dataset contains **5,268 exact duplicate rows**, representing approximately **0.97% of all records**.

These duplicates may cause double counting when calculating:

- Revenue
- Quantity sold
- Product performance
- Transaction metrics

Exact duplicate rows should therefore be removed during the data cleaning stage.

---

### 3. Cancellation and Returns

Approximately **1.66% of rows are cancellation transactions**.

Cancellation transactions are generally identified by an `InvoiceNo` beginning with `C` and negative quantities.

These transactions should not be treated as normal sales because including them in sales calculations could distort:

- Revenue
- Quantity sold
- Average Order Value
- Product performance
- Customer spending

---

### 4. Quantity Outliers

The IQR method identified **58,619 quantity outlier rows (10.82%)**.

However, investigation shows that extreme quantities are not necessarily data errors. Some may represent:

- Bulk purchases
- Large orders
- Inventory adjustments
- Damaged or unsaleable products
- Incorrect stock entries

Therefore, quantity outliers should **not be removed solely based on statistical thresholds**.

Business context must be considered before deciding whether a transaction is valid.

---

### 5. UnitPrice Outliers

The IQR method identified **39,627 UnitPrice outlier rows (7.31%)**.

Investigation revealed that many extreme prices are associated with non-product transactions such as:

- `AMAZON FEE`
- `POSTAGE`
- `Manual`
- `Adjust bad debt`

Therefore, statistical outliers in UnitPrice cannot automatically be considered invalid product prices.

---

### 6. Zero and Negative UnitPrice

There are **2,515 transactions with UnitPrice = 0** and **2 transactions with negative UnitPrice**.

The negative UnitPrice transactions are associated with `Adjust bad debt`, while many zero-price transactions have missing customer and product descriptions.

These transactions do not represent normal product sales and should be excluded from the valid sales dataset.

---

### 7. Product Identification

There are **650 StockCodes with multiple descriptions**.

This indicates that `Description` is not fully consistent across the dataset.

Therefore, `StockCode` will be treated as the primary product identifier, while `Description` will be used as supporting product information.

---

### Overall Business Interpretation

The dataset is suitable for sales performance, product analysis, customer behavior, and RFM segmentation. However, the raw dataset should not be used directly for business analysis.

The main challenge is distinguishing **valid product sales from cancellations, returns, operational adjustments, fees, and other non-standard transactions**.

Therefore, the next phase will focus on creating explicit business rules for data cleaning rather than simply removing statistical outliers.
